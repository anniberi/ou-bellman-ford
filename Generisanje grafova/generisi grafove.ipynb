```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
broj_cvorova = np.linspace(1, 1000, 1000)
# Prosjecan graf:
broj_ivica = [x*np.sqrt(x) for x in broj_cvorova]

dijkstra = broj_ivica + np.multiply(broj_cvorova, np.log2(broj_cvorova))
bf = np.multiply(broj_ivica, broj_cvorova)
fw = np.multiply(np.multiply(broj_cvorova, broj_cvorova), broj_cvorova)
```

```python
plt.figure(); plt.plot(broj_cvorova, dijkstra, color = 'blue');
plt.plot(broj_cvorova, bf, color = 'red'); plt.plot(broj_cvorova, fw, color = 'green');
plt.xlabel('Broj čvorova'); plt.ylabel('Procijenjeni broj iteracija');
plt.grid(color = 'grey', linestyle = '--', linewidth = 1)
plt.title('Usporedba brzina izvršavanja algoritama najkraćeg puta');
plt.legend(['Dijkstrin algoritam', 'Bellman-Ford algoritam', 'Floyd-Warshall algoritam']);
```
